```python
import re
def extract_n_value(filename):
    # 正则表达式解释：
    # n      : 匹配字符 'n'
    # (\d+)  : 捕获组，匹配一个或多个数字
    match = re.search(r'n(\d+)', filename)
    if match:
        return int(match.group(1)) # 返回整数类型以便正确排序
    return 0 # 如果没找到，返回0排在最后

instance_names = sorted(instance_names, key=extract_n_value, reverse=False)
```

```python
import os
import glob
import pandas as pd
from collections import deque

def get_max_time_from_csv(file_path):
    """
    读取CSV文件的最后一行以提取最大时间。
    如果文件为空，返回 0。
    """
    # 1. 检查文件是否为空
    if os.path.getsize(file_path) == 0:
        return 0.0

    try:
        # 2. 打开文件并只读取最后一行 (使用 deque 只保留最后1行)
        with open(file_path, 'r', encoding='utf-8') as f:
            # deque(f, 1) 会高效地只抓取最后一行
            last_line_gen = deque(f, maxlen=1)
            
            if not last_line_gen: # 双重检查：如果文件有大小但内容为空（如只有换行符）
                return 0.0
            
            last_line = last_line_gen[0].strip()
            
            # 3. 解析数据: {time};{fitness}
            # 假设格式总是分号分隔，且第一列是时间
            if not last_line:
                return 0.0
                
            time_str = last_line.split(';')[0]
            return float(time_str)
            
    except Exception as e:
        print(f"处理文件 {os.path.basename(file_path)} 时出错: {e}")
        return 0.0 # 出错时默认返回0，或者根据需求改为 None
```

```python


# ================= 配置路径 =================
# ===========================================

def analysis_max_times(target_path):
    method_name = os.path.basename(target_path)
    
    # 获取所有csv文件
    csv_files = glob.glob(os.path.join(target_path, "*.csv"))
    data_list = []
    
    print(f"正在处理 {len(csv_files)} 个文件...")
    
    for file in csv_files:
        max_time = get_max_time_from_csv(file)
        # 记录文件名（不含路径）和最大时间
        data_list.append({
            'filename': os.path.basename(file),
            'max_time': max_time
        })
    
    # 1. 生成 DataFrame
    df = pd.DataFrame(data_list)
    
    # 按文件名排序（可选，方便查看）
    df = df.sort_values(by='filename').reset_index(drop=True)
    
    print("--- 结果预览 (前5行) ---")
    print(df.head())
    
    print(f"\n--- 统计信息 - {method_name} ---")
    # 2. 做个统计 (均值、最大值、最小值、分位数等)
    stats = df['max_time'].describe()
    print(stats)
    
    zero_count = (df['max_time'] == 0).sum()
    print(f"\n完全无提升instance个数: {zero_count} / {len(df)}")

# 如果你想查看具体的时间分布情况
# print(df['max_time'].value_counts(bins=10))
```
